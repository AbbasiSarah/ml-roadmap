**MNIST** has 70,000 images of handwritten digits (0–9), each 28 × 28 pixels.
We'll build a small **Convolutional Neural Network (CNN)** that reads them.

💡 **Faster:** in Colab go to **Runtime → Change runtime type → T4 GPU**. (It also works on CPU, just slower.)

Run each cell in order with **Shift + Enter**.

## Step 1: Load the digits
TensorFlow and Keras are already installed in Colab. Keras downloads MNIST for us.

````python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
print(x_train.shape, x_test.shape)
````

## Step 2: An image is just numbers
Each pixel is a number from **0 (black) to 255 (white)**.

````python
plt.imshow(x_train[0], cmap="gray")
plt.title(f"Label: {y_train[0]}")
plt.show()
print(x_train[0][10:15, 10:15])   # a small piece of the grid
````

## Step 3: Prepare the images
- Divide by 255 so pixels go from **0 to 1** (small numbers help training).
- Add a **channel** dimension: grey images have 1 channel (colour images have 3).

````python
x_train = x_train / 255.0
x_test = x_test / 255.0
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)
print(x_train.shape)
````

## Step 4: Build the CNN
- **Conv2D**: small 3×3 filters slide over the image and find patterns (edges, curves).
- **MaxPooling2D**: shrinks the image and keeps the strongest signals.
- **Flatten**: turns the grid into one long list.
- **Dense(10, softmax)**: gives one score per digit; the highest wins.

````python
model = keras.Sequential([
    layers.Input((28, 28, 1)),
    layers.Conv2D(32, 3, activation="relu"),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation="relu"),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(10, activation="softmax"),
])
model.summary()
````

## Step 5: Train it
One **epoch** = one pass over all 60,000 training images.

````python
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])
model.fit(x_train, y_train, epochs=3, validation_split=0.1)
````

## Step 6: Test on new images

````python
loss, acc = model.evaluate(x_test, y_test)
print(f"Test accuracy: {acc:.2%}")
````

You should see about **99%**. Your exact number may differ slightly each time you train.

## Step 7: Predict one digit

````python
pred = model.predict(x_test[:1])
plt.imshow(x_test[0].reshape(28, 28), cmap="gray")
plt.title(f"Model says: {pred.argmax()}   (true: {y_test[0]})")
plt.show()
print("Scores for 0-9:", pred.round(3))
````
