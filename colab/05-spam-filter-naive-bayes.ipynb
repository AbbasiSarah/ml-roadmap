We have **14 messages**: 6 spam and 8 normal. Can we teach the computer to spot spam in **new** messages?

This is a small **NLP** (Natural Language Processing) project.

Run each cell in order with **Shift + Enter**.

## Step 1: Create the data
`1` = spam, `0` = normal.

````python
msgs = [
    "Congratulations! You won a free iPhone, click here",
    "URGENT: claim your prize money now",
    "Free entry! Win cash today",
    "You have won 50000 rupees, send your bank details",
    "Limited offer! Click to win a free gift",
    "Win a free holiday, reply YES now",
    "Are we still meeting at 5?",
    "Please send me the lecture notes",
    "Happy birthday! Have a great day",
    "Can you call me when you are free?",
    "The assignment is due on Monday",
    "Let's have lunch tomorrow",
    "Mom asked you to buy milk",
    "Thanks for your help today",
]
labels = [1] * 6 + [0] * 8
````

## Step 2: Turn words into numbers
Computers can't read words. `CountVectorizer` counts how many times each word appears in each message (a **bag of words**).

````python
from sklearn.feature_extraction.text import CountVectorizer

vec = CountVectorizer()
X = vec.fit_transform(msgs)
print("Number of different words:", len(vec.get_feature_names_out()))
print(vec.get_feature_names_out()[:20])
````

## Step 3: Train Naive Bayes
It learns how often each word appears in spam vs normal messages, e.g. *free, won, win, click, claim* in spam.

````python
from sklearn.naive_bayes import MultinomialNB

nb = MultinomialNB()
nb.fit(X, labels)
````

## Step 4: Test new messages

````python
new = ["You won a free gift, click now",
       "Can we meet tomorrow for lunch?",
       "Free notes for the lecture"]
print(nb.predict(vec.transform(new)))   # 1 = spam, 0 = not spam
````

## Step 5: How sure is it?
`predict_proba` gives the probability of each class: `[not spam, spam]`.

````python
for m, p in zip(new, nb.predict_proba(vec.transform(new))):
    print(f"{m:35s}  spam: {p[1]:.0%}")
````

"Free notes for the lecture" contains **free**, but the model still says **not spam**, because *notes* and *lecture* appear in normal messages.
