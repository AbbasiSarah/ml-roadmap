**The exercise:** we have data about 10 students: how many hours each one studied, and the score they got.
Can we train a model that **predicts the score** of a new student from their study hours?

We will use **Linear Regression**, the simplest machine learning model. It draws the best straight line through the data.

Run each cell in order with **Shift + Enter**.

## Step 1: Import the libraries

````python
import pandas as pd                                   # tables of data
import matplotlib.pyplot as plt                       # charts
from sklearn.linear_model import LinearRegression     # our ML model
````

## Step 2: Create the data
Each row is one student.

````python
data = {
    "hours": [1, 2, 2.5, 3, 4, 4.5, 5, 6, 7, 8],
    "score": [35, 42, 45, 50, 58, 60, 66, 72, 80, 88],
}
df = pd.DataFrame(data)
df
````

## Step 3: Look at the data
Always plot your data first. Do the points look like a straight line?

````python
plt.scatter(df["hours"], df["score"])
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("Study hours vs exam score")
plt.show()
````

## Step 4: Split into input (X) and output (y)
- **X** = what we know (hours). It needs double brackets `[[ ]]` because sklearn expects a table.
- **y** = what we want to predict (score).

````python
X = df[["hours"]]
y = df["score"]
````

## Step 5: Train the model
`.fit()` is where the model learns the best line.

````python
model = LinearRegression()
model.fit(X, y)

print("Slope:", round(model.coef_[0], 2))       # extra marks per extra hour
print("Intercept:", round(model.intercept_, 2)) # score with 0 hours
````

**What this means:** the slope is about **7.6**, so every extra hour of study adds about **7.6 marks**.
The model's formula is: `score ≈ 7.6 × hours + 26.9`

## Step 6: How good is the model?
**R²** goes from 0 to 1. Closer to 1 means the line fits the data well.

````python
print("R² score:", round(model.score(X, y), 3))
````

## Step 7: Predict a new student
A new student studied **6.5 hours**. What score will they get?

````python
new_student = pd.DataFrame({"hours": [6.5]})
prediction = model.predict(new_student)
print("Predicted score:", round(prediction[0], 1))
````

## Step 8: Draw the line

````python
plt.scatter(df["hours"], df["score"], label="Real students")
plt.plot(df["hours"], model.predict(X), color="red", label="Model's line")
plt.scatter([6.5], prediction, color="green", s=120, label="New student (6.5 h)")
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.legend()
plt.show()
````
