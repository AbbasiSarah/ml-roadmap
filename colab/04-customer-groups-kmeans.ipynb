A shop has **15 customers**. We know each one's **income** and **spending score** (0–100), but **not** their groups.
Can the computer find the groups by itself?

This is **unsupervised learning**: there is no answer column.

Run each cell in order with **Shift + Enter**.

## Step 1: Create the data

````python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

cust = pd.DataFrame({
    "income":   [15, 16, 17, 18, 20, 40, 42, 45, 47, 50, 75, 78, 80, 85, 88],
    "spending": [80, 75, 85, 78, 82, 45, 50, 48, 52, 47, 15, 20, 12, 18, 10],
})
cust
````

## Step 2: Plot the customers
Can you see the groups with your own eyes?

````python
plt.scatter(cust["income"], cust["spending"], color="gray")
plt.xlabel("Income")
plt.ylabel("Spending score")
plt.show()
````

## Step 3: Train K-Means
We choose **how many** groups (`n_clusters=3`). K-Means finds **where** they are.

````python
km = KMeans(n_clusters=3, n_init=10, random_state=0)
km.fit(cust)
print(km.labels_)
````

## Step 4: How it works
1. Put 3 centres somewhere.
2. Give each customer to the nearest centre.
3. Move each centre to the middle of its customers.
4. Repeat steps 2–3 until nothing changes.

Let's colour the customers by group:

````python
plt.scatter(cust["income"], cust["spending"], c=km.labels_)
c = km.cluster_centers_
plt.scatter(c[:, 0], c[:, 1], marker="X", s=250, color="red", label="centres")
plt.xlabel("Income")
plt.ylabel("Spending score")
plt.legend()
plt.show()
````

## Step 5: Look at the group centres

````python
print(km.cluster_centers_.round(1))
````

- **[17.2, 80]** → low income, big spenders
- **[44.8, 48.4]** → middle income, middle spending
- **[81.2, 15]** → high income, careful savers

A shop could send each group a different offer.

## Step 6: Place a new customer

````python
new = pd.DataFrame({"income": [44], "spending": [50]})
print(km.predict(new))
````

Group **0**: middle income, middle spending.

**1.** `KMeans(n_clusters=2, n_init=10, random_state=0).fit(cust).labels_`: with 2 groups, two real groups get merged; with 4, one real group gets split. 3 fits this data best.

**2.**
```python
cust2 = pd.concat([cust, pd.DataFrame({"income": [90], "spending": [85]})], ignore_index=True)
print(KMeans(n_clusters=3, n_init=10, random_state=0).fit(cust2).labels_)
```
This customer doesn't fit any group well. K-Means still forces them into one: it can't say "none of these".

**3.**
```python
scores = [KMeans(n_clusters=k, n_init=10, random_state=0).fit(cust).inertia_ for k in range(1, 7)]
plt.plot(range(1, 7), scores, marker="o")
plt.xlabel("K"); plt.ylabel("Inertia"); plt.show()
```
The curve bends (the "elbow") at **K = 3**.
